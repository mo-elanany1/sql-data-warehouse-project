# Data Structure

slicing ? changeable ? Allow Duplicates ? How we write it ?  

1- List  ==> Yes, Yes, Yes, [ Data ]
2- Tuble ==> Yes, No, Yes, ( Data )
3- Set   ==> No, Yes, No   { Data }
4- Dict  ==> Yes, Yes, No, {Key:Value}

```python
a = {1, 2, 3}
b = {3, 5}

print( a | b ) # union
print( a & b ) # intersection
print( a - b ) # difference
print( b - a ) # difference
print( a ^ b ) # symmetric difference
```

## For Lope

```python
customer = { 'mazen', 'ahmed', 'ali', 'mohamed', 'waled', 'maher' }

for i in customer:
    if i == 'maher':
        print('Done')
    else:
        print(i)
```
